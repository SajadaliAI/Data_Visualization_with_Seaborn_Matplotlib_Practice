```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
ml_students_age = np.random.randint(15,45,(100))
py_students_age = np.random.randint(18,45,(100))
```

```python
print(ml_students_age)
print(py_students_age)
```

```python
from matplotlib import style
style.use("ggplot")
bins = [15,20,25,30,35,40,45]
plt.hist([ml_students_age,py_students_age],bins,rwidth=0.8,histtype="bar",orientation="vertical",color=["m","g"],
         label=["ML Students","Py Students"])
plt.title("ML Students Histogram")
plt.xlabel("Students Catogory")
plt.ylabel("No. Students Age")
plt.legend()
```
