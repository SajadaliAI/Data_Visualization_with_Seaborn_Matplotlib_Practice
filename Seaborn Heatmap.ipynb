# Seaborn Heatmap Part 2

```python
plt.figure(figsize = (16,9))
annot_kws = {"fontsize":10,
             "fontstyle":"italic",
             "color":"k",
             "alpha":0.9,
             "rotation":"vertical",
             "verticalalignment":"center",
             "backgroundcolor":"w"}
sns.heatmap(globalWarming_df,  cmap = "coolwarm",annot = True, annot_kws = annot_kws)
plt.show()
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(globalWarming_df,  cmap = "coolwarm",annot = True ,linecolor = "k")
plt.show()
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(globalWarming_df,  cmap = "coolwarm",annot = True, linewidth = 4 ,linecolor = "k")
plt.show()
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(globalWarming_df, cbar = False)
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(globalWarming_df, cbar = False,   xticklabels = False,   yticklabels = False)
```

```python
plt.figure(figsize = (16,9)) 
cbar_kws = {"orientation":"horizontal",
             "shrink":1,
             "extend":"min",
             "extendfrac":0.1,
             "ticks":np.arange(0,22),
             "drawedges":True,}
sns.heatmap(globalWarming_df, cbar_kws = cbar_kws)
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(globalWarming_df,   xticklabels =np.arange(0,15),   yticklabels = False)
```

```python
plt.figure(figsize = (16,9))
country_code = ["USA","GBR","IND","CHN","RUS","AUS","FRA","DUE","CAN","BRA","ARG","PAK","NLP","BGD","JPN"]
sns.heatmap(globalWarming_df,   xticklabels =np.arange(0,15),   yticklabels = country_code)
```

```python
plt.figure(figsize = (17,8))
ax = sns.heatmap(globalWarming_df)
ax.set(title = "Heatmap of Who is reponsible for a global warming",
      xlabel = "(CO2 emissions metric tons per capital) per year", 
      ylabel = "Country Name")
sns.set(font_scale =3)
plt.show()
```

# Python Pandas Part 3
# Correlation Heatmap

```python
globalWarming_df.corr()
```

```python
plt.figure(figsize = (16,9))
ax = sns.heatmap(globalWarming_df.corr(), annot = True,annot_kws = {"size": 12},  linewidth = 2,)
ax.tick_params( size =10, color ="w", labelsize =10, labelcolor ="b", )
plt.title("WHo is Responsible For Global Warming",fontsize = 20)
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
cancer_dataset = load_breast_cancer()
cancer_dataset
```

```python
cancer_df = pd.DataFrame(np.c_[cancer_dataset["data"],cancer_dataset["target"]],
                        columns = np.append(cancer_dataset["feature_names"],["target"]))
cancer_df
```

```python
plt.figure(figsize = (25,25))
ax = sns.heatmap(cancer_df.corr(), annot = True,annot_kws = {"size": 14},  linewidth = 2,)
ax.tick_params( size =15, color ="w", labelsize =17, labelcolor ="b", )
plt.title("Heatmap of Cancer Patient",fontsize = 30)
plt.show()
```
