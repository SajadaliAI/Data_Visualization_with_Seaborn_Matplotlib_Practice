```python
sns.barplot(x = "day", y = "total_bill", data = tips_df, alpha = 0.9, linestyle = "-." , linewidth = 3,
            edgecolor = "g")
```

```python
ax = sns.barplot(x = "day", y = "total_bill", data = tips_df, alpha = 0.9, linestyle = "-." , linewidth = 3,
            edgecolor = "g")
ax.set(title = "Bar CHart of Total_Bill", 
      xlabel = "Days",
      ylabel = "Total_bill")
plt.show()
```

```python
plt.figure(figsize = (16,9))
sns.barplot(x = "day", y = "total_bill", data = tips_df, alpha = 1, linestyle = "-." , linewidth = 3,
            edgecolor = "k")
plt.title("Bar CHart of Total_Bill", fontsize = 15) 
plt.xlabel("Days", fontsize = 10)
plt.ylabel( "Total_bill",  fontsize = 10)
plt.show()
```
