```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
# Dataset of ("Pakistan Aritificial Inteligence Production ("PAIP")")
# Classes = ["Python","R","Artificial Intelligence","Machine Learning",Data Science]
```

```python
classes = ["Python","R","AI","ML","DS"]
class1_students =[30,10,20,25,10]
class2_students =[40,5,20,20,10]
class3_students =[35,5,30,15,15]
```

```python
plt.bar(classes,class1_students,width =0.3,align="edge",linestyle="--",edgecolor="g",color ="b",
       linewidth = 5,alpha = 0.9,label = "class 1 Students")
```

```python
style.use("ggplot")
plt.figure(figsize =(15,8))
plt.bar(classes,class1_students,width =0.3,align="edge",linestyle="--",edgecolor="g",color ="b",
       linewidth = 5,alpha = 0.9,label = "class 1 Students")
```

```python
style.use("ggplot")
plt.figure(figsize =(15,8))
plt.bar(classes,class1_students,width =0.6,align="edge",linestyle="--",edgecolor="g",color ="y",
       linewidth = 4,alpha = 0.9,label = "class 1 Students")
plt.title("Bar chart of PAIP",fontsize = 18)
plt.xlabel("NO. Students",fontsize = 15)
plt.ylabel("Classes",fontsize = 15)
plt.legend()
```

```python
style.use("ggplot")
plt.figure(figsize =(15,8))
classes_index = np.arange(len(classes))
width = 0.2
plt.bar(classes_index,class1_students,width,color ="y",label = "class 1 Students")
plt.bar(classes_index + width,class2_students,width,color ="g",label = "class 2 Students")
plt.bar(classes_index +width + width,class3_students,width,color ="m",label = "class 3 Students")
plt.title("Bar chart of PAIP",fontsize = 18)
plt.xlabel("NO. Classes",fontsize = 15)
plt.ylabel("Classes",fontsize = 15)
plt.xticks(classes_index,classes)
plt.legend()
```

```python
style.use("ggplot")
plt.figure(figsize =(15,8))
classes_index = np.arange(len(classes))
width = 0.2
plt.barh(classes_index,class1_students,width,color ="y",label = "class 1 Students")
plt.barh(classes_index + width,class2_students,width,color ="g",label = "class 2 Students")
plt.barh(classes_index +width + width,class3_students,width,color ="m",label = "class 3 Students")

plt.title("Bar chart of PAIP",fontsize = 18)
plt.xlabel("NO. Classes",fontsize = 15)
plt.ylabel("Classes",fontsize = 15)
plt.yticks(classes_index,classes)
plt.legend()
plt.show()
```
