```python
tips_df = sns.load_dataset("tips")
tips_df
```

```python
df = sns.distplot(tips_df ["total_bill"], bins = 100)
```

```python
df = sns.distplot(tips_df ["total_bill"], hist=False,)
```

```python
df = sns.distplot(tips_df ["total_bill"], rug=True,fit= norm,)
```

```python
df = sns.distplot(tips_df ["total_bill"], kde = False , fit= norm,)
```

```python
df = sns.distplot(tips_df ["total_bill"], kde = False , fit= norm, color = "r",)
```

```python
df = sns.distplot(tips_df ["total_bill"], kde = False , fit= norm, color = "r",vertical=True,
                  norm_hist=True,)
```

```python
df = sns.distplot(tips_df ["total_bill"], kde = False , fit= norm, color = "r",
                  axlabel="Sjjad Hussain",)
```

```python
df = sns.distplot(tips_df ["total_bill"], kde = False , fit= norm, color = "r",
                  axlabel="Sjjad Hussain",label = "Ahmed ALi")
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (16,9))
sns.set(style = "darkgrid")
df = sns.distplot(tips_df ["total_bill"], kde = False , fit= norm, color = "r",
             axlabel="Sjjad Hussain",label = "Ahmed ALi")
plt.title("Histogram of Total Bill", fontsize = 15)
plt.xlabel("Totab_Bill", fontsize = 10)
plt.ylabel("Tips", fontsize = 10)
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (16,9))
df = sns.distplot(tips_df ["total_bill"], hist_kws= { "color": "#DC143C", "edgecolor" : "#aaff00",
                                                    "linewidth": 5, "linestyle" : "--", "alpha" : 0.9} )   
plt.show()
```

```python
plt.figure(figsize = (16,9))
df = sns.distplot(tips_df ["total_bill"], hist_kws= { "color": "#DC143C", "edgecolor" : "#aaff00",
                                                    "linewidth": 5, "linestyle" : "--", "alpha" : 0.9},
                  
                  kde_kws= { "color": "#8e00ce", "linewidth": 9, "linestyle" : "--", "alpha" : 0.9} )   
plt.show()
```

```python
plt.figure(figsize = (16,9))
df = sns.distplot(tips_df ["total_bill"], hist_kws= { "color": "#DC143C", "edgecolor" : "#aaff00",
                                                    "linewidth": 5, "linestyle" : "--", "alpha" : 0.9},
        
                  kde_kws= { "color": "#8e00ce", "linewidth": 8, "linestyle" : "--", "alpha" : 0.9},
                 rug = True,
                 rug_kws= { "color": "#0426d0","edgecolor" : "#00dbff", "linewidth": 3, "linestyle" : "--", "alpha" : 0.9}, )   
plt.show()
```

```python
plt.figure(figsize = (16,9))
df = sns.distplot(tips_df ["total_bill"], hist_kws= { "color": "#DC143C", "edgecolor" : "#aaff00",
                           "linewidth": 5, "linestyle" : "--", "alpha" : 0.9},
                  kde = False,
                  fit = norm,
                  fit_kws= { "color": "#8e00ce", "linewidth": 8, "linestyle" : "--", "alpha" : 0.9},
                 rug = True,
                 rug_kws= { "color": "#0426d0","edgecolor" : "#00dbff", "linewidth": 3, "linestyle" : "--", "alpha" : 0.9}, )   
plt.show()
```
