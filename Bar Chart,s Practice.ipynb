```python
import matplotlib.pyplot as plt 
import numpy as np
from matplotlib import style
```

```python
# Dataset of ("Pakistan Aritificial Inteligence Production ("PAIP")")
# Classes = ["Python","R","Artificial Intelligence","Machine Learning",Data Science"]
```

```python
classes = ["Python","R","AI","ML","DS"]
class1_students = [20,30,24,29,22]
class2_students = [21,27,28,29,23]
class3_students = [23,25,27,28,30]
```

```python
style.use("ggplot")
plt.figure(figsize =(16,9))
classes_index = np.arange(len(classes))
width = 0.2
plt.bar(classes_index,class1_students,width,color = "m",label = "Class 1 Students")
plt.bar(classes_index + width,class2_students,width,color = "y",label = "Class 2 Students")
plt.bar(classes_index + width + width,class3_students,width,color = "g",label = "Class 3 Students")
plt.title("Bar Chart of PAIP Class",fontsize = 19)
plt.xlabel("NO. Students",fontsize = 15)
plt.ylabel("Classes",fontsize = 15)
plt.xticks(classes_index +width , classes)
plt.legend()
plt.show()
```
